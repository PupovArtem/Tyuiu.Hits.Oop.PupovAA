Создать базовый класс Review в C#, который будет представлять отзывы о
продуктах или услугах. На основе этого класса разработать 2-3 производных класса,

должны быть реализованы новые атрибуты и методы, а также переопределены

атрибуты, такие как Дата посещения (VisitDate).

посещения в отображение отзыва.
2. ОтзывТовара (ProductReview): Должен содержать дополнительные атрибуты,
такие как Идентификатор продукта (ProductId). Метод RateProduct() должен
быть переопределен для связывания рейтинга с конкретным продуктом.
3. ОтзывУслуги (ServiceReview) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Время начала услуги
(StartTime). Метод GetReviewDetails() должен быть переопределен для
отображения времени начала услуги вместе с другими деталями отзыва.

```


Review basicReview = new Review(1, "Отличный продукт", 4.5);
basicReview.DisplayReview();
basicReview.RateProduct(5.0);
basicReview.AddAuthor("Артём");
basicReview.VerifyReview();
Console.WriteLine(basicReview.GetReviewDetails());
Console.WriteLine();

ServiceReview serviceReview = new ServiceReview(2, "Отличный сервис", 5.0, DateTime.Now);
serviceReview.DisplayReview();
serviceReview.RateProduct(4.0);
Console.WriteLine(serviceReview.GetReviewDetails());
Console.WriteLine();

ProductReview productReview = new ProductReview(3, "Качественный товар", 4.8, "PROD-123");
productReview.DisplayReview();
productReview.RateProduct(3.5);
Console.WriteLine(productReview.GetReviewDetails());
Console.WriteLine();

DetailedServiceReview detailedReview = new DetailedServiceReview(4,"Превосходное обслуживание",4.9,DateTime.Now.AddDays(-3),DateTime.Now.AddHours(-5));

detailedReview.DisplayReview();
detailedReview.RateProduct(4.7);
Console.WriteLine(detailedReview.GetReviewDetails());
Console.WriteLine();

SpecialReview specialReview = new SpecialReview(5, "Хороший товар", 5.0);
specialReview.DisplayReview();
specialReview.AddComment("Отличный отзыв!");
specialReview.ShareReview();

double averageRating = (basicReview.Rating + serviceReview.Rating + productReview.Rating + detailedReview.Rating) / 4;

Console.WriteLine($"\nОбщий рейтинг: {averageRating:F1}");

Console.WriteLine("\n Перегрузка методов");

basicReview.AddLike(3);
basicReview.RateProduct(4.5, "Хороший продукт");
basicReview.RateProduct(4);

serviceReview.SetEmployee("Иван", 5);
serviceReview.SetAddress("Москва");

productReview.RecommendProduct(false);
productReview.SetBrand("Example");

detailedReview.CalculateDuration(true);

specialReview.AddComment("Рекомендую!", "Артём");
specialReview.ShareReview("Telegram");

// Демонстрация дополнительных методов
basicReview.PublishReview();
specialReview.FeatureReview();



List<Review> reviews = new List<Review>
{
    basicReview,
    serviceReview,
    productReview,
    detailedReview,
    specialReview
};

foreach (Review review in reviews)
{
    review.DisplayReview();
    Console.WriteLine(review.GetReviewDetails());
    Console.WriteLine($"Положительный отзыв: {review.IsPositive()}");
    Console.WriteLine();
}


Console.WriteLine("Generic-классы");

ReviewRepository<Review> repository = new ReviewRepository<Review>();

foreach (Review review in reviews)
{
    repository.Add(review);
}

Console.WriteLine($"Всего отзывов: {repository.Count}");
repository.DisplayAll();

Review foundReview = repository.FindById(3);

if (foundReview != null)
{
    Console.WriteLine("\nНайденный отзыв:");
    Console.WriteLine(foundReview.GetReviewDetails());
}
ReviewResponse<string> response = new ReviewResponse<string>( true,"Обработка отзывов завершена","Успешно");

response.Display();


public class Review
{
    private int _reviewId;
    private string _text;
    private double _rating;

    public string Author { get; set; }
    public int Likes { get; set; }
    public bool IsVerified { get; set; }

    // Добавленные атрибуты
    public DateTime CreatedAt { get; set; }
    public int Views { get; set; }
    public bool IsPublished { get; set; }

    public int ReviewId
    {
        get { return _reviewId; }
        set { _reviewId = value; }
    }

    public string Text
    {
        get { return _text; }
        set { _text = value; }
    }

    public double Rating
    {
        get { return _rating; }
        set { _rating = ValidateRating(value); }
    }

    public Review(int reviewId, string text, double rating)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = ValidateRating(rating);
        Author = "Аноним";
        Likes = 0;
        IsVerified = false;

        // Инициализация добавленных атрибутов
        CreatedAt = DateTime.Now;
        Views = 0;
        IsPublished = true;
    }

    protected double ValidateRating(double rating)
    {
        if (rating < 1 || rating > 5)
            throw new ArgumentException("Рейтинг должен быть от 1 до 5!");

        return rating;
    }

    public virtual void DisplayReview()
    {
        Views++;
        Console.WriteLine($"Отзыв #{ReviewId}: \"{Text}\"");
    }

    public virtual void RateProduct(double newRating)
    {
        Rating = ValidateRating(newRating);
        Console.WriteLine($"Рейтинг: {Rating} из 5");
    }

    // Перегрузка RateProduct
    public void RateProduct(double newRating, string comment)
    {
        Rating = ValidateRating(newRating);
        Console.WriteLine($"Рейтинг: {Rating} из 5. Комментарий: {comment}");
    }

    // Перегрузка RateProduct с целым числом
    public void RateProduct(int stars)
    {
        Rating = ValidateRating(stars);
        Console.WriteLine($"Установлено звёзд: {stars}");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID: {ReviewId}, Текст: {Text}, Рейтинг: {Rating}, " +
               $"Автор: {Author}, Лайки: {Likes}, Проверен: {IsVerified}, " +
               $"Дата создания: {CreatedAt}, Просмотры: {Views}, " +
               $"Опубликован: {IsPublished}";
    }

    public void AddLike()
    {
        Likes++;
    }

    // Перегрузка AddLike
    public void AddLike(int count)
    {
        if (count < 0)
            throw new ArgumentException("Количество лайков не может быть отрицательным!");

        Likes += count;
    }

    public void VerifyReview()
    {
        if (!IsVerified)
        {
            IsVerified = true;
        }
        else
        {
            IsVerified = false;
        }
    }

    public void AddAuthor(string author)
    {
        Author = author;
    }

    // Добавленный метод
    public void PublishReview()
    {
        IsPublished = true;
        Console.WriteLine("Отзыв опубликован.");
    }

    // Виртуальный метод для полиморфизма
    public virtual bool IsPositive()
    {
        return Rating >= 4;
    }
}

public class ServiceReview : Review
{
    private DateTime _visitDate;

    public string EmployeeName { get; set; }
    public int ServiceDuration { get; set; }
    public bool WasAppointment { get; set; }

    // Добавленные атрибуты
    public string ServiceAddress { get; set; }
    public double ServiceQuality { get; set; }
    public bool IsRepeatCustomer { get; set; }

    public DateTime VisitDate
    {
        get { return _visitDate; }
        set { _visitDate = value; }
    }

    public ServiceReview(int reviewId, string text, double rating, DateTime visitDate)
        : base(reviewId, text, rating)
    {
        VisitDate = visitDate;
        EmployeeName = "Не указан";
        ServiceDuration = 0;
        WasAppointment = false;

        ServiceAddress = "Не указан";
        ServiceQuality = 5.0;
        IsRepeatCustomer = false;
    }

    public void SetEmployee(string name)
    {
        EmployeeName = name;
    }

    public void SetDuration(int minutes)
    {
        ServiceDuration = minutes;
    }

    // Перегрузка SetEmployee
    public void SetEmployee(string name, int quality)
    {
        if (quality < 1 || quality > 5)
            throw new ArgumentException("Качество должно быть от 1 до 5!");

        EmployeeName = name;
        ServiceQuality = quality;
    }

    public void SetAddress(string address)
    {
        ServiceAddress = address;
    }

    public override void DisplayReview()
    {
        Views++;
        Console.WriteLine($"Отзыв #{ReviewId}: \"{Text}\" Дата посещения: {VisitDate:dd.MM.yyyy}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $", Дата посещения: {VisitDate:dd.MM.yyyy}, " +
               $"Сотрудник: {EmployeeName}, Качество услуги: {ServiceQuality}";
    }

    public override bool IsPositive()
    {
        return Rating >= 4 && ServiceQuality >= 3;
    }
}

public class ProductReview : Review
{
    private string _productId;

    public string ProductName { get; set; }
    public string Category { get; set; }
    public bool Recommended { get; set; }

    // Добавленные атрибуты
    public decimal ProductPrice { get; set; }
    public string Brand { get; set; }
    public int PurchaseCount { get; set; }

    public string ProductId
    {
        get { return _productId; }
        set { _productId = value; }
    }

    public ProductReview(int reviewId, string text, double rating, string productId)
        : base(reviewId, text, rating)
    {
        ProductId = productId;
        ProductName = "Не указан";
        Category = "Без категории";
        Recommended = true;

        ProductPrice = 0;
        Brand = "Не указан";
        PurchaseCount = 0;
    }

    public void SetProductName(string name)
    {
        ProductName = name;
    }

    public void RecommendProduct()
    {
        Recommended = true;
    }

    // Перегрузка RecommendProduct
    public void RecommendProduct(bool recommended)
    {
        Recommended = recommended;
    }

    public void SetBrand(string brand)
    {
        Brand = brand;
    }

    public override void RateProduct(double newRating)
    {
        Rating = ValidateRating(newRating);
        Console.WriteLine($"Продукт с ID {ProductId} получил рейтинг: {Rating} из 5");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $", ID товара: {ProductId}, Категория: {Category}, " +
               $"Бренд: {Brand}, Рекомендован: {Recommended}";
    }

    public override bool IsPositive()
    {
        return Rating >= 4 && Recommended;
    }
}

public class DetailedServiceReview : ServiceReview
{
    private DateTime _startTime;

    public DateTime EndTime { get; set; }
    public string ServiceType { get; set; }
    public bool IncludedMaterials { get; set; }

    // Добавленные атрибуты
    public string CustomerRequirements { get; set; }
    public double Price { get; set; }
    public string ResultDescription { get; set; }

    public DateTime StartTime
    {
        get { return _startTime; }
        set { _startTime = value; }
    }

    public DetailedServiceReview(
        int reviewId,
        string text,
        double rating,
        DateTime visitDate,
        DateTime startTime)
        : base(reviewId, text, rating, visitDate)
    {
        StartTime = startTime;
        EndTime = startTime.AddHours(1);
        ServiceType = "Стандартная услуга";
        IncludedMaterials = false;

        CustomerRequirements = "Не указаны";
        Price = 0;
        ResultDescription = "Не указан";
    }

    public void SetServiceType(string type)
    {
        ServiceType = type;
    }

    public void CalculateDuration()
    {
        TimeSpan duration = EndTime - StartTime;
        Console.WriteLine($"Продолжительность: {duration.TotalMinutes:F0} минут");
    }

    // Перегрузка CalculateDuration
    public double CalculateDuration(bool showResult)
    {
        TimeSpan duration = EndTime - StartTime;

        if (showResult)
        {
            Console.WriteLine($"Продолжительность: {duration.TotalMinutes:F0} минут");
        }

        return duration.TotalMinutes;
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $", Время начала: {StartTime:HH:mm}, Тип услуги: {ServiceType}, " +
               $"Стоимость: {Price}";
    }
}

public interface ICommentable
{
    void AddComment(string comment);
}

public interface IShareable
{
    void ShareReview();
}

public class SpecialReview : Review, ICommentable, IShareable
{
    public string Comment { get; set; }
    public int Shares { get; set; }
    public string Platform { get; set; }

    // Добавленные атрибуты
    public string Moderator { get; set; }
    public bool IsFeatured { get; set; }
    public DateTime? FeaturedUntil { get; set; }

    public SpecialReview(int reviewId, string text, double rating)
        : base(reviewId, text, rating)
    {
        Comment = "";
        Shares = 0;
        Platform = "Не указана";

        Moderator = "Не назначен";
        IsFeatured = false;
        FeaturedUntil = null;
    }

    public void AddComment(string comment)
    {
        Comment = comment;
        Console.WriteLine($"Комментарий: {Comment}");
    }

    // Перегрузка AddComment
    public void AddComment(string comment, string author)
    {
        Comment = comment;
        Console.WriteLine($"Комментарий от {author}: {Comment}");
    }

    public void ShareReview()
    {
        Shares++;
        Console.WriteLine($"Отзыв опубликован {Shares} раз.");
    }

    // Перегрузка ShareReview
    public void ShareReview(string platform)
    {
        Platform = platform;
        Shares++;
        Console.WriteLine($"Отзыв опубликован на платформе {Platform}");
    }

    public void SetPlatform(string platform)
    {
        Platform = platform;
    }

    public void FeatureReview()
    {
        IsFeatured = true;
        Console.WriteLine("Отзыв отмечен как избранный.");
    }

    public override void DisplayReview()
    {
        base.DisplayReview();
        Console.WriteLine($"Специальный отзыв. Избранный: {IsFeatured}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() +
               $", Платформа: {Platform}, Количество публикаций: {Shares}, " +
               $"Избранный: {IsFeatured}";
    }
}

// Generic-класс для хранения отзывов
public class ReviewRepository<T> where T : Review
{
    private List<T> reviews = new List<T>();

    public void Add(T review)
    {
        reviews.Add(review);
    }

    public T FindById(int id)
    {
        return reviews.Find(r => r.ReviewId == id);
    }

    public void DisplayAll()
    {
        foreach (T review in reviews)
        {
            review.DisplayReview();
        }
    }

    public int Count
    {
        get { return reviews.Count; }
    }
}

// Второй generic-класс
public class ReviewResponse<T>
{
    public bool Success { get; set; }
    public string Message { get; set; }
    public T Data { get; set; }

    public ReviewResponse(bool success, string message, T data)
    {
        Success = success;
        Message = message;
        Data = data;
    }

    public void Display()
    {
        Console.WriteLine($"Успех: {Success}");
        Console.WriteLine($"Сообщение: {Message}");
        Console.WriteLine($"Данные: {Data}");
    }
}
```
