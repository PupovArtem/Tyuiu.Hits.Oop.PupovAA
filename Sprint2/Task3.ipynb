<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Review в C#, который будет представлять отзывы о
продуктах или услугах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

1. ОтзывОбслуживания (ServiceReview): Должен содержать дополнительные
атрибуты, такие как Дата посещения (VisitDate).
Метод DisplayReview() должен быть переопределен для включения даты
посещения в отображение отзыва.
2. ОтзывТовара (ProductReview): Должен содержать дополнительные атрибуты,
такие как Идентификатор продукта (ProductId). Метод RateProduct() должен
быть переопределен для связывания рейтинга с конкретным продуктом.
3. ОтзывУслуги (ServiceReview) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Время начала услуги
(StartTime). Метод GetReviewDetails() должен быть переопределен для
отображения времени начала услуги вместе с другими деталями отзыва.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
Review basicReview = new Review(1, "Отличный продукт", 4.5);
basicReview.DisplayReview();
basicReview.RateProduct(5.0);
basicReview.AddAuthor("Артём");
basicReview.VerifyReview();
Console.WriteLine(basicReview.GetReviewDetails());
Console.WriteLine();

ServiceReview serviceReview = new ServiceReview(2, "Отличный сервис", 5.0, DateTime.Now);
serviceReview.DisplayReview();
serviceReview.RateProduct(4.0);
Console.WriteLine(serviceReview.GetReviewDetails());
Console.WriteLine();

ProductReview productReview = new ProductReview(3, "Качественный товар", 4.8, "PROD-123");
productReview.DisplayReview();
productReview.RateProduct(3.5);
Console.WriteLine(productReview.GetReviewDetails());
Console.WriteLine();

DetailedServiceReview detailedReview = new DetailedServiceReview(4,"Превосходное обслуживание",4.9,DateTime.Now.AddDays(-3),DateTime.Now.AddHours(-5));

detailedReview.DisplayReview();
detailedReview.RateProduct(4.7);
Console.WriteLine(detailedReview.GetReviewDetails());
Console.WriteLine();

SpecialReview specialReview = new SpecialReview(5, "Хороший товар", 5.0);
specialReview.DisplayReview();
specialReview.AddComment("Отличный отзыв!");
specialReview.ShareReview();

double averageRating = (basicReview.Rating + serviceReview.Rating + productReview.Rating + detailedReview.Rating) / 4;
Console.WriteLine($"\nОбщий рейтинг: {averageRating:F1}");

public class Review
{
    private int _reviewId;
    private string _text;
    private double _rating;

    public string Author { get; set; }
    public int Likes { get; set; }
    public bool IsVerified { get; set; }

    public int ReviewId
    {
        get { return _reviewId; }
        set { _reviewId = value; }
    }

    public string Text
    {
        get { return _text; }
        set { _text = value; }
    }

    public double Rating
    {
        get { return _rating; }
        set { _rating = value; }
    }

    public Review(int reviewId, string text, double rating)
    {
        ReviewId = reviewId;
        Text = text;
        Rating = ValidateRating(rating);
        Author = "Аноним";
        Likes = 0;
        IsVerified = false;
    }

    protected double ValidateRating(double rating)
    {
        if (rating < 1 || rating > 5)
            throw new ArgumentException("Рейтинг должен быть от 1 до 5!");
        return rating;
    }

    public virtual void DisplayReview()
    {
        Console.WriteLine($"Отзыв #{ReviewId}: \"{Text}\"");
    }

    public virtual void RateProduct(double newRating)
    {
        Rating = ValidateRating(newRating);
        Console.WriteLine($"Рейтинг: {Rating} из 5");
    }

    public virtual string GetReviewDetails()
    {
        return $"ID: {ReviewId}, Текст: {Text}, Рейтинг: {Rating}, Автор: {Author}, Лайки: {Likes}, Проверен: {IsVerified}";
    }

    public void AddLike()
    {
        Likes++;
    }

    public void VerifyReview()
    {
        if (!IsVerified)
        {
            IsVerified=true;
        }
        else
        {
            IsVerified=false;
        }
    }
    public void AddAuthor(string author)
    {
        Author= author;
    }
}

public class ServiceReview : Review
{
    private DateTime _visitDate;

    public string EmployeeName { get; set; }
    public int ServiceDuration { get; set; }
    public bool WasAppointment { get; set; }

    public DateTime VisitDate
    {
        get { return _visitDate; }
        set { _visitDate = value; }
    }

    public ServiceReview(int reviewId, string text, double rating, DateTime visitDate)
        : base(reviewId, text, rating)
    {
        VisitDate = visitDate;
        EmployeeName = "Не указан";
        ServiceDuration = 0;
        WasAppointment = false;
    }

    public void SetEmployee(string name)
    {
        EmployeeName = name;
    }

    public void SetDuration(int minutes)
    {
        ServiceDuration = minutes;
    }

    public override void DisplayReview()
    {
        Console.WriteLine($"Отзыв #{ReviewId}: \"{Text}\" Дата посещения: {VisitDate:dd.MM.yyyy}");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", Дата посещения: {VisitDate:dd.MM.yyyy}, Сотрудник: {EmployeeName}";
    }
}

public class ProductReview : Review
{
    private string _productId;

    public string ProductName { get; set; }
    public string Category { get; set; }
    public bool Recommended { get; set; }

    public string ProductId
    {
        get { return _productId; }
        set { _productId = value; }
    }

    public ProductReview(int reviewId, string text, double rating, string productId)
        : base(reviewId, text, rating)
    {
        ProductId = productId;
        ProductName = "Не указан";
        Category = "Без категории";
        Recommended = true;
    }

    public void SetProductName(string name)
    {
        ProductName = name;
    }

    public void RecommendProduct()
    {
        Recommended = true;
    }

    public override void RateProduct(double newRating)
    {
        Rating = ValidateRating(newRating);
        Console.WriteLine($"Продукт с ID {ProductId} получил рейтинг: {Rating} из 5");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", ID товара: {ProductId}, Категория: {Category}";
    }
}

public class DetailedServiceReview : ServiceReview
{
    private DateTime _startTime;

    public DateTime EndTime { get; set; }
    public string ServiceType { get; set; }
    public bool IncludedMaterials { get; set; }

    public DateTime StartTime
    {
        get { return _startTime; }
        set { _startTime = value; }
    }

    public DetailedServiceReview(int reviewId, string text, double rating, DateTime visitDate, DateTime startTime)
        : base(reviewId, text, rating, visitDate)
    {
        StartTime = startTime;
        EndTime = startTime.AddHours(1);
        ServiceType = "Стандартная услуга";
        IncludedMaterials = false;
    }

    public void SetServiceType(string type)
    {
        ServiceType = type;
    }

    public void CalculateDuration()
    {
        TimeSpan duration = EndTime - StartTime;
        Console.WriteLine($"Продолжительность: {duration.TotalMinutes:F0} минут");
    }

    public override string GetReviewDetails()
    {
        return base.GetReviewDetails() + $", Время начала: {StartTime:HH:mm}, Тип услуги: {ServiceType}";
    }
}

public interface ICommentable
{
    void AddComment(string comment);
}

public interface IShareable
{
    void ShareReview();
}

public class SpecialReview : Review, ICommentable, IShareable
{
    public string Comment { get; set; }
    public int Shares { get; set; }
    public string Platform { get; set; }

    public SpecialReview(int reviewId, string text, double rating)
        : base(reviewId, text, rating)
    {
        Comment = "";
        Shares = 0;
        Platform = "Не указана";
    }

    public void AddComment(string comment)
    {
        Comment = comment;
        Console.WriteLine($"Комментарий: {Comment}");
    }

    public void ShareReview()
    {
        Shares++;
        Console.WriteLine($"Отзыв опубликован {Shares} раз.");
    }

    public void SetPlatform(string platform)
    {
        Platform = platform;
    }
}

```
